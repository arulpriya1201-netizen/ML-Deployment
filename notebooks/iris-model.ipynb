```python
import pandas as pd
import pickle
from sklearn.linear_model import LogisticRegression
df = pd.read_csv(r"D:\Users\91638\Downloads\iris_data.csv")

```

```python
x = df.iloc[:, :-1].values
y = df.iloc[:, -1].values
```

```python
lr=LogisticRegression(random_state=0)
model=lr.fit(x,y)
```

```python
with open('iris-model.pkl', 'wb') as f:
    pickle.dump(model, f)
```
